### 3. Fact Table (`FactFlights`)
Contains granular flight measures (duration in minutes and hours, departure hour, arrival hour) and foreign keys referencing dimension tables.

````python
import os
import pandas as pd
import numpy as np

PROCESSED_DIR = os.path.join("..", "data", "processed")
df_flights = pd.read_csv(os.path.join(PROCESSED_DIR, "final_flights.csv"))
print(f"FactFlights Table: {df_flights.shape[0]} rows, {df_flights.shape[1]} columns")
display(df_flights.head(3))
````

### 4. Dimension Tables
- **`DimAirline`:** Unique airlines (`SpiceJet`, `Air India`, `Vistara`, `IndiGo`).
- **`DimRoute`:** Unique route corridors (`DEL → BOM`, `BOM → CCU`, etc.).
- **`DimTime`:** Operational time shifts (`Night`, `Morning`, `Afternoon`, `Evening`) and overnight flags.

````python
dim_airline = df_flights[["airline"]].drop_duplicates().sort_values("airline").reset_index(drop=True)
dim_route = df_flights[["route", "source", "destination"]].drop_duplicates().sort_values("route").reset_index(drop=True)
dim_time = df_flights[["departure_period", "departure_hour", "overnight_flag"]].drop_duplicates().sort_values("departure_hour").reset_index(drop=True)

print(f"DimAirline: {len(dim_airline)} rows")
print(f"DimRoute:   {len(dim_route)} rows")
print(f"DimTime:    {len(dim_time)} rows")
````

### 6. Average Flight Duration KPI
Calculated mean flight duration in minutes and hours across valid records.

````python
valid_flights = df_flights[df_flights["duration_status"] == "Valid"]
avg_min = valid_flights["flight_duration_minutes"].mean()
avg_hrs = avg_min / 60.0
print(f"Average Duration (Minutes) : {avg_min:.2f} mins")
print(f"Average Duration (Hours)   : {avg_hrs:.2f} hrs")
print(f"Min Duration               : {valid_flights['flight_duration_minutes'].min():.2f} mins")
print(f"Max Duration               : {valid_flights['flight_duration_minutes'].max():.2f} mins")
````
