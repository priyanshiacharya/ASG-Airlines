### 3. Data Ingestion
In this section, we import required libraries, verify file paths, and load the Excel workbook.

```python
import os
import pandas as pd

# Define raw data file path
RAW_DATA_PATH = os.path.join("..", "data", "raw", "UseCase - Airlines.xlsx")

# Verify file existence
if os.path.exists(RAW_DATA_PATH):
    print(f"✓ Raw dataset found at: {os.path.abspath(RAW_DATA_PATH)}")
else:
    print(f"✗ Error: Raw dataset not found at {RAW_DATA_PATH}")
```

```python
# Load Excel Workbook and detect available sheets
excel_file = pd.ExcelFile(RAW_DATA_PATH, engine="openpyxl")
sheet_names = excel_file.sheet_names

print(f"Total Sheets Detected: {len(sheet_names)}")
print(f"Sheet Names: {sheet_names}")
```

### 4. Sheet Inspection
We load each individual sheet into a dictionary of DataFrames and inspect row and column counts.

```python
sheets_data = {}

for sheet in sheet_names:
    df = pd.read_excel(excel_file, sheet_name=sheet)
    sheets_data[sheet] = df
    print(f"Sheet: '{sheet:<12}' | Rows: {df.shape[0]:<6} | Columns: {df.shape[1]:<4}")
```

### 5. Basic Schema Inspection
Below we preview the schema, column data types, non-null counts, and top sample rows for each table.

```python
for sheet_name, df in sheets_data.items():
    print(f"\n{'='*50}")
    print(f"SCHEMA SUMMARY: {sheet_name.upper()}")
    print(f"{'='*50}")
    print(f"Columns: {list(df.columns)}\n")
    print(df.info())
    display(df.head(3))
```
